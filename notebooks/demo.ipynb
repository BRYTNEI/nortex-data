# NORTEX Data en San Lorenzo

Instala `.[notebook]` y configura las credenciales en `.env`. Las fuentes de clima y satélite usan enero de 2025. FAOSTAT usa un año y GAEZ consulta periodos climáticos.

```python
import matplotlib.pyplot as plt

from nortex_data import get_dataset

AOI = 'parcela_01'
```

## NASA POWER

```python
try:
    power = get_dataset("power", AOI, '2025-01-01', '2025-01-31')
except Exception as error:
    print(error)
else:
    display(power.head())
    for (variable, unit), group in power.groupby(['variable', 'unit']):
        dates = group.date.notna().any()
        if not dates:
            continue
        _, ax = plt.subplots(figsize=(9, 2.5))
        for product, series in group.groupby('product'):
            ax.plot(series.date, series.value, marker='.', linestyle='none', label=product)
        ax.set(title=variable, ylabel=unit)
        ax.legend()
        plt.show()
```
